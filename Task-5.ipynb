```python
#checking missing values
import pandas as pd
data={
    "Name":["Lahari","sindhu",None,"gayathri"],
"Marks":[80,90,None,95]
}
df=pd.DataFrame(data)
print(df.isnull())
print(df.isnull().sum())
    
```

```python
#drop missing values
import pandas as pd
data={
    "Name":["Lahari","sindhu",None,"gayathri"],
"Marks":[80,90,None,95]
}
df=pd.DataFrame(data)
df.dropna(inplace=True)
print(df)
```

```python
#Remove duplicate rows
import pandas as pd
data={
    "Name":["Lahari","sindhu","sindhu","gayathri"],
"Marks":[80,90,90,95]
}
df=pd.DataFrame(data)
print(df.drop_duplicates())
```
