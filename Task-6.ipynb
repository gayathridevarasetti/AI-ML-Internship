```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,30,70]
plt.plot(x,y)
plt.show()
```

```python
#simple line chart
import matplotlib.pyplot as plt
x=[40,30,4,8]
y=[80,20,30,70]
plt.plot(x,y,color="red")
plt.title("Line Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#green line with star mark
import matplotlib.pyplot as plt
x=[1,4,7,0]
y=[67,7,8,90]
plt.plot(x,y,color="green",marker="*")
plt.show()
```

```python
#two lines graph
import matplotlib.pyplot as plt
x=[1,2,3,4]
y1=[10,20,30,40]
y2=[30,60,90,15]
plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

```python
#Bar chart
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[5,7,9,3]
plt.bar(x,y)
plt.show()
```

```python
#scatter plot with red dots
import matplotlib.pyplot as plt
x=[20,30,40]
y=[2,5,8]
plt.scatter(x,y,color="red")
plt.show()
```

```python
#histogram
import matplotlib.pyplot as plt
marks=[100,90,64,80,70]
plt.hist(marks)
plt.title("Marks Distribution")
plt.show()
```

```python
#pie chart
import matplotlib.pyplot as plt
lang=["Python","Java","C","C++"]
students=[30,20,25,25]
plt.pie(students,labels=lang)
plt.show()
```
